```python
from numpy import random
# Generate  2D array on random numbers
x=random.rand(5,2)
print(x)
# print a digit of choice in this 1 D array
y=random.choice([1,3,5,8,10])
print(y)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# displot ising searbon,histogram plot
sns.displot([0,1,2,3,4,5],kind="kde")
plt.show()
```
